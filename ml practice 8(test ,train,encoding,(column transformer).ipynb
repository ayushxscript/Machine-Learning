```python
import pandas as pd

df = pd.read_csv(r"D:\datasets-for-ml-main\datasets-for-ml-main\covid_toy.csv")

df
```

```python
from sklearn.model_selection import train_test_split

x = df.drop('has_covid', axis=1)
y = df['has_covid']

x_train, x_test, y_train, y_test = train_test_split(
    x, y, test_size=0.2, random_state=42
)
```

```python
x_train
```

```python
from sklearn.impute import SimpleImputer

si = SimpleImputer(strategy='mean')

# Training data
x_train_fever = si.fit_transform(x_train[['fever']])

# Testing data
x_test_fever = si.transform(x_test[['fever']])

x_train_fever.shape
```

```python
from sklearn.preprocessing import OrdinalEncoder

oe = OrdinalEncoder(categories=[['Mild', 'Strong']])

# Training data
x_train_cough = oe.fit_transform(x_train[['cough']])

# Testing data
x_test_cough = oe.transform(x_test[['cough']])
```

```python
from sklearn.preprocessing import OneHotEncoder

ohe = OneHotEncoder(drop='first', sparse_output=False)

# Training data
x_train_gender_city = ohe.fit_transform(x_train[['gender', 'city']])

# Testing data
x_test_gender_city = ohe.transform(x_test[['gender', 'city']])

# Display data
x_train_gender_city
```

```python
# Training data
x_train_age = x_train.drop(
    columns=['gender', 'fever', 'cough', 'city']
).values

# Testing data
x_test_age = x_test.drop(
    columns=['gender', 'fever', 'cough', 'city']
).values

# Display data
x_train_age
```

```python
#column transformer
from sklearn.compose import ColumnTransformer

transformer = ColumnTransformer(transformers=[
    # fill missing values in the 'fever' column (mean, median or mode)
    ('tnf1', SimpleImputer(), ['fever']),
    # encode the 'cough' column in a fixed order
    ('tnf2', OrdinalEncoder(categories=[['Mild', 'Strong']]), ['cough']),
    # one-hot encode gender and city
    ('tnf3', OneHotEncoder(sparse_output=False, drop='first'), ['gender', 'city'])
], remainder='passthrough')  # the rest of the columns stay the same


```
